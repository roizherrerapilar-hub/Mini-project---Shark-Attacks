# Día 2 - Data Cleaning Process
**Objetivo:** Limpiar el dataset de ataques de tiburón para identificar los 3 lugares menos peligrosos donde abrir una franquicia de buceo.

**Secciones:**
1. Manejo de valores nulos
2. Manejo de duplicados
3. Manipulación de cadenas de texto
4. Uso de regex

6. Verificación final


```python
#1. Valores nulos
#Hemos eliminado en la primera parte todas las columnas y filas con valores nulos. 
```

```python
# Count the number of null values in each column
df.isna().sum()
```

```python
#2. Valores duplicados
df = df.drop_duplicates()
df.head()
```

```python
df.shape 

#Vemos que el dataframe se ha reducido de (39196, 255) a (7087, 15)
```

```python
#3. Manipulacion de strings
#Poner los nombres de las columnas en minuscula
df.columns = df.columns.str.lower()
df.head()
```

```python
#Columnas que consideramos importantes para nuestro proyecto: type, country, state, location, activity, age, injury, species. 
#4. Eliminacion de las columnas innecesarias:
df = df.drop(columns=['date', 'year', 'type', 'name', 'sex', 'fatal y/n', 'time', 'source'])
df.head()
```

```python
#Visualización de columnas
df = df.drop(columns=df.columns[15:])
df.columns = df.columns.str.lower()
print(df.columns.tolist())
```

```python
#Eliminar duplicados de filas iguales
#Resetear el índice
df = df.drop_duplicates().reset_index(drop=True)
```

```python
print(f'Shape de partida: {df.shape}')
print(f'Columnas: {df.columns.tolist()}')
df.head()
```

```python
#Visualizar los valores de la columna age
age_values = df['age'].tolist()
print(age_values)
```

```python
# Función regex: extrae el primer número entero que encuentre
# \d+ = uno o más numeros consecutivos
import re
def extraer_edad(valor):
    match = re.search(r'\d+', str(valor))
    return float(match.group()) if match else None
```

```python
df['age'] = df['age'].apply(extraer_edad)
```

```python
age_values = df['age'].tolist()
print(age_values)
```

```python
# Eliminar filas donde no se pudo extraer ningún número
df = df.dropna(subset=['age'])

# Filtrar edades imposibles
df = df[(df['age'] >= 1) & (df['age'] <= 100)]
```

```python
print(f'Shape tras limpiar age: {df.shape}')
print(f'Rango de edades: {int(df["age"].min())} - {int(df["age"].max())} años')
```

```python
# Convertir age a entero (ya sin nulos ni valores imposibles)
df['age'] = df['age'].astype(int)
df.head()
```

```python
# Crear rangos de edad con pd.cut()
bins   = [0, 17, 25, 35, 50, 65, 100]
labels = ['<18', '18-25', '26-35', '36-50', '51-65', '65+']
```

```python
# Crear nueva columna 'age_range' con los rangos definidos
df['age_range'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)
```

```python
# Visualizar la distribución de los rangos de edad
print('Distribución por rango de edad:')
print(df['age_range'].value_counts().sort_index())
```

```python
# Verificar que no quedan nulos
print('Nulos restantes:')
print(df.isna().sum())
```

#comentario
-Solo quedan 9 nulos en country y 6 en injury. Los eliminamos porque son columnas críticas para nuestro análisis:

```python
#Dado que necesitamos saber el dato de en que countries hubo ataques eliminamos los datos de los que no tenemos esa informacion
# Sin país no podemos ubicar el ataque, sin injury no sabemos si fue fatal
df = df.dropna(subset=['country', 'injury'])
print(f'Shape final: {df.shape}')
```

```python
#Para conseguir un ranking necesitamos una tasa de fatalidad y mortalidad con la cual poder descartar paises de forma rotunda y saber con cuales quedarnos
# Crear columna booleana 'fatal'
# True  → el texto de injury contiene la palabra 'fatal'
# False → cualquier otro caso

df['fatal'] = df['injury'].str.contains('fatal', case=False, na=False)

print('Distribución fatal vs no fatal:')
print(df['fatal'].value_counts())
print(f'\n% ataques fatales: {df["fatal"].mean()*100:.1f}%')
```

#Comentario 
-De 4017 ataques registrados, 683 fueron fatales — un 17%.

# Día 3 - Finalizar estructura y limpieza de datos
**Objetivo:** Generar una mejor visual de los datos para mayor precisión del resultado.

**Secciones:**
1. Finalizar limpieza y formateo de datos 
2. Estructurar los datos para mayos sostenibilidad del analisis
3. Uso de agragaciones y tecnicas de filtrado.
4. Uso de graficos de apoyo

```python
#Unificar los nombres de los paises. 
df['country'] = df['country'].astype(str).str.strip().str.upper()  

limpieza_country = {
    'USA': 'UNITED STATES',
    'US': 'UNITED STATES',
    'U.S.A.': 'UNITED STATES'
}

df['country'] = df['country'].replace(limpieza_country)
```

```python
#limpiar valores de columnas de country y injury
df['state'] = df['state'].str.replace('NSW', 'New South Wales', case=False)
df['state'] = df['state'].str.replace('Between New Ireland & New Britain', 'New Ireland / Britain', case=False)
df['state'] = df['state'].str.title()

df['injury'] = df['injury'].astype(str).str.strip()
df['injury'] = df['injury'].str.replace(r'\bR\b', 'right', case=False, regex=True)
df['injury'] = df['injury'].str.replace(r'\bL\b', 'left', case=False, regex=True)
df['injury'] = df['injury'].str.lower().str.capitalize()

print(df['injury'].unique()[:10])
```

```python
df.head()
```

```python
# Limpieza de la columna 'activity'
def activity_category(activity):
    if pd.isna(activity):
        return 'Unknown'
    activity = str(activity).lower()
    if 'div' in activity or 'snorkel'in activity:
        return 'Diving'
    elif 'surf' in activity:
        return 'Surfing' 
    elif 'swim' in activity:
        return 'Swimming'
    elif 'fish' in activity: 
        return 'Fishing'
    elif 'boat' in activity or 'kayak' in activity:
        return 'Boating' 
    else: 
        return 'Other'
```

```python
# Aplicar la función para obtener los nombres de las actividades y total 
df['activity_category'] = df['activity'].apply(activity_category)
df['activity_category'].value_counts()    
```

```python
df.head()
```

```python
#definir el riesgo. 
if 'fatal y/n' in df.columns: 
    df['fatal'] = df['fatal y/n'].astype(str).str.strip().str.upper().map({'Y': 1, 'N': 0})
else:
    df['fatal'] = df['injury'].str.contains('fatal', case=False, na=False).astype(int)

print(df['fatal'].value_counts(dropna=False))
```

```python
df.head()
```

```python
#Numero total de ataques por pais
country_total_attacks = df.groupby('country').size()
country_total_attacks
```

```python
#Numero total de ataques fatales por pais
country_fatal_attacks = df.groupby('country')['fatal'].sum()
country_fatal_attacks
```

```python
#proporción de ataques que terminan en muerte en cada país
fatal_rate = country_fatal_attacks / country_total_attacks
fatal_rate
```

```python
#unir en una misma tabla los datos de total de ataques, ataques fatales y tasa de fatalidad por pais
country_summary = pd.concat(
    [country_total_attacks, country_fatal_attacks, fatal_rate], axis=1)
country_summary.columns = ['total_attacks', 'fatal_attacks', 'fatal_rate']
country_summary
```

```python
#filtramos para quedarnos con los paises que tienen suficientes casos y el analisis tiene sentido.
#Ordenar de menos a mayor
country_summary = country_summary[country_summary['total_attacks'] >= 10]
country_summary = country_summary.sort_values('fatal_rate')
country_summary.head(10).round(2)
```
